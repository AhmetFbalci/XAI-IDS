```python
import numpy as np
import pandas as pd

# 1) Label'ı çıkar
df_clean = df.drop(columns="Label")

# 2) inf değerleri temizle
df_clean = df_clean.replace([np.inf, -np.inf], np.nan)

# 3) NaN varsa temizle (isteğe bağlı ama önerilir)
df_clean = df_clean.dropna()

# 4) sadece sayısal sütunlar
numeric_df = df_clean.select_dtypes(include=["number"])

# 5) variance hesapla
variances = numeric_df.var()

# 6) variance = 0 olan feature'lar
zero_var_cols = variances[variances == 0].index

# 7) çıkarılan feature'ları yazdır
print("Silinen feature'lar (variance = 0):")
for col in zero_var_cols:
    print("-", col)

# 8) dataset'ten çıkar
df_clean = df_clean.drop(columns=zero_var_cols)

print("\nYeni dataset shape:", df_clean.shape)


```

```python
df_final = df_clean.copy()
df_final["Label"] = df["Label"]
```

```python
corr_matrix = df_final.corr(numeric_only=True)
print(corr_matrix)

```

```python
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# çiftleri al
corr_pairs = upper.stack()

# 0.9'dan büyük olanlar
high_corr = corr_pairs[abs(corr_pairs) > 0.99]

print(high_corr)
```
